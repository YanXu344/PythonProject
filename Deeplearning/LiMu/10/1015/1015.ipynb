```python
X = torch.arange(12,dtype = torch.float).reshape(3,4)
Y = torch.tensor([[1,2,3,4],[4,2,1,3],[3,2,1,3]])
before = id(Y)
display(id(Y))
Y = Y+X
id(Y) == before
display(id(Y))
id(Y) == before
```

```python
Z = torch.zeros_like(Y)
print(id(Z))
Z[:] = X
print(id(Z))
Z[:] = X+Y 
print(id(Z))
```
