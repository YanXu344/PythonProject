```python
import os
path = os.path.join('..','data')
os.makedirs(path,exist_ok = True)
data_file = os.path.join(path,'House_csv')
with open (data_file,'w') as file:
    file.write('NumRooms,Alley,HouseAge,Garage,Area,Price\n')
    file.write('NA,Pave,10,NA,85,127500\n')
    file.write('2,NA,15,NA,70,106000\n')
    file.write('4,Grvl,NA,Yes,120,178100\n')
    file.write('NA,NA,20,NA,95,140000\n')
    file.write('3,Pave,12,NA,105,160000\n')
    file.write('5,Grvl,8,No,150,210000\n')
    file.write('2,NA,18,NA,80,135000\n')
    file.write('4,Pave,25,NA,130,190000\n')
    file.write('3,Grvl,NA,NA,110,165000\n')
```

```python
import pandas as pd
data = pd.read_csv(data_file)
data
```

* 数据清洗

```python
def drop_col(m):
    col = m.isna().sum().idxmax()
    m = m.drop(columns = [col])
    return m  
```

```python
inputs,outputs = data.iloc[:,0:5],data.iloc[:,5]
inputs,outputs
```

```python
inputs = drop_col(inputs)
display(inputs)
```

```python
inputs = inputs.fillna(inputs.mean(numeric_only=True))
print(inputs)
```

```python
inputs = pd.get_dummies(inputs, columns=['Alley'], dummy_na=True, dtype=int)
display(inputs)
```

```python
import torch
x = torch.tensor(inputs.to_numpy(dtype='float32'))
y = torch.tensor(outputs.to_numpy(dtype='float32'))
display(x)
display(y)
```
